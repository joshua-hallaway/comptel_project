```python
import gammalib
import ctools
import contextlib
```

```python
with open("gammalib_help.txt", "w") as f:
    with contextlib.redirect_stdout(f):
        for x in dir(gammalib):
            if x.startswith("GCOM"):
                cls = getattr(gammalib, x)
                print(f"=== {x} ===")
                help(cls)
                print("\n")
```
